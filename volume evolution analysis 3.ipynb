```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd

df = pd.read_excel("bitcoin market features table 10.xlxs.xlsx")
```

```python
df[["trade_date", "close_price"]].head(10)
```

```python
df["trade_date"] = pd.to_datetime(df["trade_date"])
```

```python
df = df.sort_values("trade_date")
```

```python
plt.figure(figsize=(15,6))

plt.plot(df["trade_date"], df["volume"])

plt.title("Bitcoin Daily Trading Volume", fontsize=18)

plt.xlabel("Year")

plt.ylabel("Volume")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

```python
yearly_volume = (
    df.groupby(df["trade_date"].dt.year)["volume"]
      .mean()
)

plt.figure(figsize=(12,6))

plt.bar(yearly_volume.index.astype(str), yearly_volume.values)

plt.title("Average Annual Trading Volume", fontsize=18)

plt.xlabel("Year")

plt.ylabel("Average Volume")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    df["trade_date"],
    df["close_price"],
    label="Closing Price",
    linewidth=2
)

plt.plot(
    df["trade_date"],
    df["volume"] / 1000000,
    label="Volume (Millions)",
    alpha=0.7
)

plt.title("Bitcoin Price vs Trading Volume", fontsize=18)

plt.xlabel("Year")

plt.ylabel("Price / Volume")

plt.legend()

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```
